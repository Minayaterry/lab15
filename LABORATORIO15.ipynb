```python
# Paso 2: Importar librerías
import chromadb
from chromadb.config import Settings
from sentence_transformers import SentenceTransformer
import pprint
```

```python
import chromadb

# Crear cliente con configuración por defecto
client = chromadb.Client()

# Crear colección llamada “Tecsup”
collection = client.get_or_create_collection(name="Tecsup")
```

```python

# Paso 4: Agregar información personal (tú y 3 compañeros)
documentos = [
    "Me llamo Ronald, estudio Big Data y me gusta la estadística.",
    "Mi compañera Andrea estudia Inteligencia Artificial y vive en Arequipa.",
    "Luis es estudiante de Ciencia de Datos y le interesa la visualización de datos.",
    "Sandra es de Trujillo, le gusta el machine learning y es buena en Python."
]

metadatos = [
    {"nombre": "Ronald"},
    {"nombre": "Andrea"},
    {"nombre": "Luis"},
    {"nombre": "Sandra"}
]

collection.add(
    documents=documentos,
    metadatas=metadatos,
    ids=["1", "2", "3", "4"]
)
```

```python
# Paso 5: Crear prompt para hacer consulta
prompt = "¿Quién es bueno en Python?"

resultado_1 = collection.query(
    query_texts=[prompt],
    n_results=2
)

print(" Resultado usando motor por defecto de chromadb:")
pprint.pprint(resultado_1)
```

OTRO MOTOR DE EMBEDDING

```python
# Paso 6: Crear nueva colección usando SentenceTransformer
modelo = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
# Generar embeddings personalizados
embeddings = modelo.encode(documentos).tolist()
```

```python
# Crear otra colección y guardar con embeddings manuales
collection_alt = client.get_or_create_collection(name="Tecsup_SBERT")
collection_alt.add(
    documents=documentos,
    metadatas=metadatos,
    ids=["1", "2", "3", "4"],
    embeddings=embeddings
)
```

```python

# Codificar el mismo prompt y buscar
prompt_embedding = modelo.encode([prompt]).tolist()

resultado_2 = collection_alt.query(
    query_embeddings=prompt_embedding,
    n_results=2
)

print("\n Resultado usando SentenceTransformer (SBERT):")
pprint.pprint(resultado_2)
```
